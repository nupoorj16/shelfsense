```python
import pandas as pd

reviews = pd.read_json(
    "../data/raw/Beauty_and_Personal_Care.jsonl.gz", lines=True, nrows=100_000
)
reviews.head()
```

```python
print("rows:", len(reviews))
print("unique users:", reviews["user_id"].nunique())
print("unique products:", reviews["parent_asin"].nunique())
print(reviews["rating"].value_counts().sort_index())
print("dates:", reviews["timestamp"].min(), "to", reviews["timestamp"].max())
```

```python
per_user = reviews.groupby("user_id").size()
per_user.describe()
```
